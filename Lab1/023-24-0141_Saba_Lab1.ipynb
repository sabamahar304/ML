```python
#Task_2
# Approach: Check whether the number is divisible by any number from 2 to n-1.
def is_prime(n):

    if n < 2:
        return False

    for i in range(2, n):

        if n % i == 0:
            return False

    return True

number = 13
print(is_prime(number))

number = 48
print(is_prime(number))
```

```python
#Task_3
# Approach: Use a loop and continuously add the previous two numbers to generate the sequence.
n = 10

a = 0
b = 1

print("Fibonacci sequence:")

for i in range(n):

    print(a, end=" ")
    next = a + b
    a = b
    b = next
```

```python
#Task_4
# Approach: Go through the list and add a number only if it is not already in the result list.
def remove_dup(numbers):

    result = []

    for i in numbers:

        if i not in result:
            result.append(i)

    return result

numbers = [10, 20, 10, 30, 20, 40, 30]

print(remove_dup(numbers))
```

```python
#Task_5
# Approach: Use *args to accept any number of values and multiply them using a loop.
def multiply(*args):

    result = 1

    for number in args:
        result = result * number

    return result

print(multiply(2, 3, 4))
```

```python
#Task_6
# Approach: Go through each character and use a dictionary to count how many times it appears.

text = "happy"
frequency = {}

for character in text:
    if character in frequency:
        frequency[character] += 1
    else:
        frequency[character] = 1

print(frequency)
```

```python
#Task_7
# Approach: Compare each employee's salary with the current highest salary and update it when needed.
employees = [
    {"name": "Ali", "department": "IT", "salary": 73400},
    {"name": "Sana", "department": "HR", "salary": 85020},
    {"name": "Uzair", "department": "IT", "salary": 75300}
]

HSalary = employees[0]

for i in employees:

    if i["salary"] > HSalary["salary"]:
        HSalary = i


print("Employee with highest salary:")
print(HSalary)
```

```python
#Task_8
# Approach: Use lambda with filter() to keep only numbers whose remainder after division by 2 is not zero.
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

odd_numbers = list(
    filter(lambda x: x % 2 != 0, numbers)
)

print(odd_numbers)
```

```python
#Task_13
# Approach: Use * for element-wise multiplication and @ for matrix multiplication to compare their results.
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

E_wise = A * B

M_mul = A @ B

print("Element-wise multiplication:\n" , E_wise )

print("Matrix multiplication:\n" , M_mul)
```

```python
#Task_15
# Approach: Apply the given normalization formula to scale all values
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

minimum = np.min(numbers)
maximum = np.max(numbers)

normalized = (numbers - minimum) / (maximum - minimum)

print(normalized)
```

```python
#Task_16
# Approach: Use axis=1 to calculate the total and average marks for each student row.
import numpy as np

marks = np.array([
    [80, 70, 90],
    [60, 75, 65],
    [90, 85, 95],
    [70, 80, 75],
    [88, 92, 84]
])

total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)

print("Total marks:")
print(total)

print("Average marks:")
print(average)
```

```python
#Task_17
# Approach: Use np.where() to replace even numbers with -1 while keeping odd numbers unchanged.
import numpy as np

numbers = np.array([1, 2, 3, 4, 5, 6, 7, 8])

result = []
for num in numbers:
    if num % 2 == 0:
        result.append(-1)
    else:
        result.append(int(num))

print(result)
```

```python
#Task_18
# Approach: Create a DataFrame from student information and use describe() for statistical summary.
import pandas as pd

data = {
    "name": [
        "Ali", "Sara", "Omar", "Hira",
        "Ahmed", "Zain", "Ayesha", "Usman"
    ],

    "section": [
        "A", "A", "B", "B",
        "A", "B", "A", "B"
    ],

    "marks": [
        85, 92, 67, 78,
        45, 88, 73, 56
    ]
}

df = pd.DataFrame(data)

print(df)

print("\nDescriptive statistics:")
print(df.describe())
```

```python
#Task_19
# Approach: Load the CSV, count missing values, and fill missing numeric values with their column mean.
import pandas as pd

df = pd.read_csv("Students Social Media Addiction.csv")

print("Missing values before filling:")
print(df.isna().sum())

numeric_columns = df.select_dtypes(include="number").columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

print("\nMissing values after filling:")
print(df.isna().sum())
```

```python
#Task_20
# Approach: Use boolean filtering with .loc to select students whose marks are below 50.
below_50 = df.loc[df["marks"] < 50, ["name", "marks"]]

print(below_50)
```

```python
#Task_21
# Approach: Group students by section and calculate the mean and maximum marks for each section.
section_stats = df.groupby("section")["marks"].agg(
    ["mean", "max"]
)

print(section_stats)
```

```python
#Task_22
# Approach: Merge the student and attendance DataFrames using student_id, then filter attendance below 75%.
import pandas as pd

students = pd.DataFrame({
    "student_id": [1, 2, 3, 4],
    "name": ["Ali", "Sara", "Omar", "Hira"]
})

attendance = pd.DataFrame({
    "student_id": [1, 2, 3, 4],
    "attendance": [80, 70, 90, 60]
})

merged = pd.merge(
    students,
    attendance,
    on="student_id"
)

low_attendance = merged[
    merged["attendance"] < 75
]

print(low_attendance)
```
